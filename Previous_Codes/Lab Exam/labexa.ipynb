```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df=pd.read_csv('hospital_doctor_data.csv')

```

```python
print("First 5 rows of the dataset:")
print(df.head())
```

```python
#Box Plot of monthly salary across departments
plt.figure(figsize=(8,5))
sns.boxplot(x="Department", y="Monthly_Salary", data=df)

plt.title("Distribution of Monthly Salary Across Departments")
plt.xlabel("Department")
plt.ylabel("Monthly Salary")

plt.show()
```

```python
#Correlation Matrix and Heatmap
numerical_data = df[[
    "Doctor_Experience",
    "Patients_Per_Day",
    "Monthly_Salary",
    "Satisfaction_Score"
]]

# Compute correlation matrix
corr_matrix = numerical_data.corr()

print("\nCorrelation Matrix:")
print(corr_matrix)

# Plot heatmap
plt.figure(figsize=(7,5))
sns.heatmap(corr_matrix,
            annot=True,
            cmap="coolwarm",
            fmt=".2f")

plt.title("Correlation Heatmap of Numerical Features")
plt.show()
```

```python
#Pairplot of Numerical Variables
sns.pairplot(df,
             vars=["Doctor_Experience",
                   "Patients_Per_Day",
                   "Monthly_Salary",
                   "Satisfaction_Score"],
             hue="Department")

plt.show()
```
